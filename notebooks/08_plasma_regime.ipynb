**Notebook for flagging plasma regimes for given data.**

Define function to look at geometry to output summary for given probe for some time range:

```python
import numpy as np
import pyspedas
from pyspedas import get_data, time_double, time_string

R_E = 6371.0

def regime_look(trange, probe='c', v_sw=400.0):
    """Print geometry and median ion moments for one probe over a time range."""

    #load probe positions (state) and ion moments (esa) into spedas store
    pyspedas.projects.themis.state(probe=probe, trange=trange, time_clip=True)
    pyspedas.projects.themis.esa(probe=probe, trange=trange, level='l2', time_clip=True)

    pos = get_data(f'th{probe}_pos_gse').y #position in km, GSE
    a = np.arctan(30.0 / v_sw) #aberration angle due to windsock effect
    tail = np.array([-np.cos(a), np.sin(a), 0.0]) #aberrated tail axis
    r = np.linalg.norm(pos, axis=1)
    ang = np.degrees(np.arccos(np.clip(pos @ tail / r, -1, 1)))  #0 = down the tail

    n = get_data(f'th{probe}_peif_density').y
    V = get_data(f'th{probe}_peif_velocity_gse').y
    T = get_data(f'th{probe}_peif_avgtemp').y

    print(trange[0], f'  r = {np.median(r) / R_E:5.1f} R_E   angle from tail axis = {np.median(ang):5.1f} deg')
    print(f'   n = {np.nanmedian(n):6.2f} cm^-3   |V| = {np.nanmedian(np.linalg.norm(V, axis=1)):6.0f} km/s' f'   Vx = {np.nanmedian(V[:, 0]):6.0f} km/s   T = {np.nanmedian(T):7.1f} eV')
```

```python
regime_look(['2013-12-04/00:00', '2013-12-04/00:30']) #Sawyer pass
regime_look(['2013-12-17/12:00', '2013-12-17/13:00']) #near full Moon probably be in the tail
```
